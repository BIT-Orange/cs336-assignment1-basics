```python
import random

from cs336_basics.tokenizer import Tokenizer
from pathlib import Path

random.seed(42)

tinystories_tokenizer = Tokenizer.from_files(
    vocab_filepath=Path("output/tinystories/vocab.txt"),
    merges_filepath=Path("output/tinystories/merges.txt"),
    special_tokens=["<|endoftext|>"]
)

tinystories = Path("../data/TinyStoriesV2-GPT4-valid.txt").read_text(encoding="utf-8")
tinystories_sample = random.sample(tinystories.splitlines(), 1000)
tinystories_total_tokens = sum(len(tinystories_tokenizer.encode(line)) for line in tinystories_sample)
tinystories_total_bytes = sum(len(line.encode("utf-8")) for line in tinystories_sample)


tinystories_ratio = tinystories_total_bytes / tinystories_total_tokens

print(f"Total tokens: {tinystories_total_tokens}")
print(f"Total bytes: {tinystories_total_bytes}")
print(f"Average bytes per token: {tinystories_ratio:.4f}")


```
